```python
import numpy as np
arr = np.array([10, 20, 30, 40, 50])

print( arr)
print( arr.ndim)
print( arr.shape)
print( arr.size)
print(arr.dtype)
```

```python
arr2 = np.random.randint(0, 51, size=(4, 5))


print(arr2)

print( arr2.itemsize)
print( arr2.nbytes)

print(arr2[arr2 % 5 == 0])
```

```python

arr3 = np.arange(1, 13).reshape(3, 4)


print(arr3)

print( arr3.shape[0])
print( arr3.shape[1])

reversed_columns = arr3[:, ::-1]

print(reversed_columns)
```

```python
arr4 = np.arange(1, 13).reshape(3, 4)

print(arr4)
view = arr4[0:2, 0:2]
print(view)
view[0, 0] = 100

print(view)


print(arr4)

```
